# Exploratory Data Analysis (EDA)

## Objetivo

Esta etapa tem como objetivo explorar o comportamento das variáveis do dataset IEEE-CIS Fraud Detection, identificar padrões associados às fraudes e gerar insights que apoiem a Engenharia de Dados, a seleção de atributos e o desenvolvimento dos modelos de Machine Learning.

```python
# Bibliotecas

import warnings
import matplotlib.pyplot as plt
import pandas as pd

warnings.filterwarnings("ignore")

pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", 100)
```

```python
# Caminho dos arquivos
DATA_PATH = "../data/raw"

# Carregar apenas o conjunto de treinamento
train_transaction = pd.read_csv(
    f"{DATA_PATH}/train_transaction.csv"
)

print(train_transaction.shape)
train_transaction.head()
```

## Distribuição da variável alvo

Antes de explorar as variáveis explicativas, é importante compreender a distribuição da variável alvo (`isFraud`), verificando a proporção entre transações legítimas e fraudulentas.

```python
fraud_dist = (
    train_transaction["isFraud"]
    .value_counts()
    .rename_axis("Classe")
    .to_frame("Quantidade")
)

fraud_dist["Percentual (%)"] = (
    fraud_dist["Quantidade"]
    / fraud_dist["Quantidade"].sum()
    * 100
).round(2)

display(fraud_dist)
```

```python
fraud_dist["Quantidade"].plot(kind="bar")

plt.title("Distribuição da Variável Alvo (isFraud)")
plt.xlabel("Classe")
plt.ylabel("Quantidade")

plt.show()
```

### Insight

O conjunto de treinamento apresenta forte desbalanceamento entre as classes.

Apenas **3,5%** das transações correspondem a fraudes, enquanto **96,5%** representam transações legítimas.

Esse comportamento confirma que se trata de um problema real de detecção de fraude e reforça a necessidade de utilizar métricas apropriadas (como Precision, Recall, F1-Score e ROC-AUC) durante a avaliação dos modelos de Machine Learning.

## Análise da Taxa de Fraude por Faixa de Valor

As análises anteriores permitiram compreender a distribuição da variável **TransactionAmt** e comparar seus principais indicadores entre transações legítimas e fraudulentas.

Entretanto, médias, medianas e boxplots nem sempre respondem a uma pergunta importante para o negócio:

> **A probabilidade de uma transação ser fraudulenta aumenta conforme cresce o seu valor?**

Para responder a essa questão, as transações serão agrupadas em faixas de valor. Em seguida, será calculada a **taxa de fraude** em cada faixa, permitindo comparar o risco relativo entre diferentes intervalos de valores.

Essa abordagem facilita a interpretação dos resultados e pode indicar padrões úteis para a Engenharia de Atributos, para o treinamento dos modelos de Machine Learning e para a construção de indicadores em dashboards analíticos.

```python
# Criar as faixas

bins = [0, 50, 100, 250, 500, float("inf")]

labels = [
    "≤ R$ 50",
    "R$ 50-100",
    "R$ 100-250",
    "R$ 250-500",
    "> R$ 500"
]

train_transaction["Faixa_TransactionAmt"] = pd.cut(
    train_transaction["TransactionAmt"],
    bins=bins,
    labels=labels,
    include_lowest=True
)

# Os intervalos são fechados à direita (right=True),
# por exemplo: (50, 100] inclui 100, mas não inclui 50.
```

```python
# Taxa de fraude por faixa

fraud_rate = (
    train_transaction
    .groupby("Faixa_TransactionAmt", observed=True)["isFraud"]
    .agg(
        Transacoes="count",
        Fraudes="sum",
        Taxa_Fraude=lambda x: round(x.mean() * 100, 2)
    )
)

display(fraud_rate)
```

```python
# Gráfico

plt.figure(figsize=(9,5))

plt.bar(
    fraud_rate.index.astype(str),
    fraud_rate["Taxa_Fraude"]
)

plt.title("Taxa de Fraude por Faixa de Valor da Transação")
plt.xlabel("Faixa de TransactionAmt")
plt.ylabel("Taxa de Fraude (%)")

plt.xticks(rotation=15)

plt.show()
```

# Análise da variável ProductCD

A variável **ProductCD** representa a categoria do produto associado à transação.

Nesta etapa serão analisados:

- distribuição das transações por categoria;
- quantidade de fraudes por categoria;
- taxa de fraude por categoria.

O objetivo é identificar se determinadas categorias apresentam risco significativamente maior e se essa variável possui potencial para contribuir com os modelos de detecção de fraude.

```python
# Distribuição das categorias

product_dist = (
    train_transaction["ProductCD"]
    .value_counts()
    .rename_axis("ProductCD")
    .to_frame("Transações")
)

product_dist["Percentual (%)"] = (
    product_dist["Transações"]
    / product_dist["Transações"].sum()
    * 100
).round(2)

display(product_dist)
```

```python
# Gráfico de volume de transações

plt.figure(figsize=(8,5))

plt.bar(
    product_dist.index,
    product_dist["Transações"]
)

plt.title("Distribuição de Transações por ProductCD")
plt.xlabel("Categoria do Produto")
plt.ylabel("Quantidade de Transações")

plt.show()
```

```python
# Taxa de fraude por categoria

product_fraud = (
    train_transaction
    .groupby("ProductCD", observed=True)["isFraud"]
    .agg(
        Transacoes="count",
        Fraudes="sum",
        Taxa_Fraude=lambda x: round(x.mean() * 100, 2)
    )
    .sort_values("Taxa_Fraude", ascending=False)
)

display(product_fraud)
```

```python
# Gráfico da taxa de fraude

plt.figure(figsize=(8,5))

plt.bar(
    product_fraud.index,
    product_fraud["Taxa_Fraude"]
)

plt.title("Taxa de Fraude por ProductCD")
plt.xlabel("Categoria do Produto")
plt.ylabel("Taxa de Fraude (%)")

plt.show()
```

### Insight

A variável **ProductCD** apresenta diferenças expressivas de risco entre as categorias de produto.

A categoria **C** possui a maior taxa de fraude (**11,69%**), significativamente acima da taxa média do dataset (**3,5%**). Em contraste, a categoria **W**, embora concentre a maior parte das transações, apresenta a menor taxa de fraude (**2,04%**).

Esse resultado demonstra que o tipo de produto está fortemente associado à probabilidade de fraude e representa uma variável com elevado potencial preditivo.

A análise também evidencia que **volume de transações não implica maior risco**. Categorias menos frequentes podem apresentar probabilidade de fraude substancialmente superior, o que reforça a importância de utilizar taxas de fraude, e não apenas quantidades absolutas, na avaliação de risco.

## Interação entre ProductCD e TransactionAmt

A análise anterior mostrou que a categoria **C** apresenta a maior taxa de fraude do dataset.

Nesta etapa, será investigado se essa maior incidência de fraude está associada a valores mais elevados de transação ou se o risco permanece alto independentemente do valor da transação.

A comparação será realizada por meio das estatísticas de **TransactionAmt** para cada categoria de produto, separando transações legítimas e fraudulentas.

```python
# Estatísticas por categoria e fraude

product_amount = (
    train_transaction
    .groupby(["ProductCD", "isFraud"], observed=True)["TransactionAmt"]
    .agg(
        Transacoes="count",
        Media="mean",
        Mediana="median",
        Q1=lambda x: x.quantile(0.25),
        Q3=lambda x: x.quantile(0.75),
        Maximo="max"
    )
    .round(2)
)

display(product_amount)
```

```python
# Comparação visual

plt.figure(figsize=(8,5))

(
    train_transaction
    .groupby(["ProductCD", "isFraud"], observed=True)["TransactionAmt"]
    .median()
    .unstack()
    .plot(kind="bar")
)

plt.title("Mediana de TransactionAmt por ProductCD e Classe")
plt.xlabel("ProductCD")
plt.ylabel("Mediana de TransactionAmt")
plt.legend(["Não fraude", "Fraude"])

plt.tight_layout()
plt.show()
```

### Insight

A análise conjunta de **ProductCD** e **TransactionAmt** mostra que a elevada taxa de fraude da categoria **C** não pode ser explicada apenas pelo valor das transações.

Embora a categoria C apresente a maior taxa de fraude do dataset (**11,69%**), suas transações possuem valores relativamente baixos. A mediana das transações fraudulentas é de **R$ 34,80**, muito próxima da mediana das transações legítimas (**R$ 30,78**).

Em contraste, as categorias **H**, **R** e **W** apresentam aumento significativo do valor das transações fraudulentas em relação às transações legítimas.

Esses resultados indicam que diferentes categorias de produto possuem perfis de fraude distintos. Na categoria C, o risco parece estar associado à própria natureza da categoria, enquanto em outras categorias o valor da transação exerce influência mais relevante.

Essa evidência reforça a importância de utilizar **ProductCD** e **TransactionAmt** de forma conjunta durante a Engenharia de Atributos e o treinamento dos modelos de Machine Learning.

# Análise da variável TransactionDT

A variável **TransactionDT** representa o instante relativo em que a transação ocorreu.

Embora o dataset não forneça a data de calendário correspondente, é possível extrair padrões temporais importantes, como:

* evolução das transações ao longo do tempo;
* variação da taxa de fraude;
* comportamento por hora do dia;
* diferenças entre transações legítimas e fraudulentas.

O objetivo desta etapa é identificar componentes temporais que possam contribuir para a Engenharia de Atributos e para os modelos de detecção de fraude.


```python
# Estatísticas básicas

train_transaction["TransactionDT"].describe()
```

```python
# Converter para dias relativos

train_transaction["TransactionDay"] = (
    train_transaction["TransactionDT"] // (24 * 60 * 60)
).astype(int)

train_transaction[["TransactionDT", "TransactionDay"]].head()
```

```python
# Volume de transações por dia

daily_transactions = (
    train_transaction
    .groupby("TransactionDay")["isFraud"]
    .agg(
        Transacoes="count",
        Fraudes="sum"
    )
)

daily_transactions["Taxa_Fraude"] = (
    daily_transactions["Fraudes"]
    / daily_transactions["Transacoes"]
    * 100
)

display(daily_transactions.head())
```

```python
# Evolução das transações

plt.figure(figsize=(12,5))

plt.plot(
    daily_transactions.index,
    daily_transactions["Transacoes"]
)

plt.title("Volume de Transações por Dia Relativo")
plt.xlabel("Dia relativo")
plt.ylabel("Quantidade de transações")

plt.tight_layout()
plt.show()
```

```python
# Evolução da taxa de fraude

plt.figure(figsize=(12,5))

plt.plot(
    daily_transactions.index,
    daily_transactions["Taxa_Fraude"]
)

plt.title("Taxa de Fraude por Dia Relativo")
plt.xlabel("Dia relativo")
plt.ylabel("Taxa de fraude (%)")

plt.tight_layout()
plt.show()
```

```python
# Hora do dia

train_transaction["TransactionHour"] = (
    (train_transaction["TransactionDT"] // 3600) % 24
).astype(int)

hourly_fraud = (
    train_transaction
    .groupby("TransactionHour")["isFraud"]
    .agg(
        Transacoes="count",
        Fraudes="sum"
    )
)

hourly_fraud["Taxa_Fraude"] = (
    hourly_fraud["Fraudes"]
    / hourly_fraud["Transacoes"]
    * 100
)

display(hourly_fraud)
```

```python
# Taxa de fraude por hora

plt.figure(figsize=(10,5))

plt.plot(
    hourly_fraud.index,
    hourly_fraud["Taxa_Fraude"],
    marker="o"
)

plt.title("Taxa de Fraude por Hora do Dia")
plt.xlabel("Hora do dia")
plt.ylabel("Taxa de fraude (%)")

plt.xticks(range(24))

plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

### Insight

A variável **TransactionHour** apresentou o padrão temporal mais expressivo observado até o momento.

A taxa de fraude aumenta significativamente entre **4h e 9h**, atingindo o pico às **7h**, com **10,61%** das transações classificadas como fraudulentas. Após as **10h**, observa-se uma queda acentuada do risco, com os menores valores registrados entre **13h e 17h**.

O pico de fraude ocorre em um período de **baixo volume de transações**, indicando que o aumento da taxa não é explicado pelo maior número de operações, mas por uma mudança no perfil das transações realizadas nesse horário.

Esse resultado demonstra que o componente temporal possui elevado potencial preditivo e justifica a criação da variável **TransactionHour** como atributo permanente para as etapas de Engenharia de Atributos e modelagem de Machine Learning.


# Análise das variáveis DeviceType e DeviceInfo

As variáveis **DeviceType** e **DeviceInfo** estão presentes na tabela `train_identity` e fornecem informações sobre o dispositivo utilizado na transação.

Nesta etapa serão analisados:

* distribuição dos tipos de dispositivo;
* taxa de fraude por tipo de dispositivo;
* principais dispositivos utilizados;
* relação entre dispositivo e risco de fraude.

O objetivo é verificar se o contexto do dispositivo contribui para diferenciar transações legítimas e fraudulentas.


```python
# Carregar informações de identidade
train_identity = pd.read_csv(
    f"{DATA_PATH}/train_identity.csv"
)

print(train_identity.shape)
train_identity.head()
```

```python
# Integrar identidade ao conjunto principal

eda_identity = train_transaction.merge(
    train_identity[["TransactionID", "DeviceType", "DeviceInfo"]],
    on="TransactionID",
    how="left"
)

eda_identity[["TransactionID", "DeviceType", "DeviceInfo"]].head()
```

```python
# Distribuição de DeviceType

device_type_dist = (
    eda_identity["DeviceType"]
    .fillna("Desconhecido")
    .value_counts()
    .rename_axis("DeviceType")
    .to_frame("Transacoes")
)

device_type_dist["Percentual (%)"] = (
    device_type_dist["Transacoes"]
    / device_type_dist["Transacoes"].sum()
    * 100
).round(2)

display(device_type_dist)
```

```python
# Taxa de fraude por tipo de dispositivo

device_type_fraud = (
    eda_identity
    .assign(DeviceType=eda_identity["DeviceType"].fillna("Desconhecido"))
    .groupby("DeviceType", observed=True)["isFraud"]
    .agg(
        Transacoes="count",
        Fraudes="sum",
        Taxa_Fraude=lambda x: round(x.mean() * 100, 2)
    )
    .sort_values("Taxa_Fraude", ascending=False)
)

display(device_type_fraud)
```

```python
# Gráfico

plt.figure(figsize=(8,5))

plt.bar(
    device_type_fraud.index.astype(str),
    device_type_fraud["Taxa_Fraude"]
)

plt.title("Taxa de Fraude por Tipo de Dispositivo")
plt.xlabel("DeviceType")
plt.ylabel("Taxa de Fraude (%)")

plt.tight_layout()
plt.show()
```

## Principais dispositivos

A variável **DeviceInfo** possui alta cardinalidade, com grande variedade de dispositivos e sistemas operacionais.

Em vez de analisar todas as categorias, serão avaliados os dispositivos mais frequentes e suas respectivas taxas de fraude.


```python
# Top 15 dispositivos

top_devices = (
    eda_identity["DeviceInfo"]
    .fillna("Desconhecido")
    .value_counts()
    .head(15)
)

display(top_devices)
```

```python
# Taxa de fraude dos principais dispositivos

top_device_names = top_devices.index.tolist()

device_info_fraud = (
    eda_identity[
        eda_identity["DeviceInfo"]
        .fillna("Desconhecido")
        .isin(top_device_names)
    ]
    .assign(DeviceInfo=eda_identity["DeviceInfo"].fillna("Desconhecido"))
    .groupby("DeviceInfo", observed=True)["isFraud"]
    .agg(
        Transacoes="count",
        Fraudes="sum",
        Taxa_Fraude=lambda x: round(x.mean() *100, 2)
    )
    .sort_values("Taxa_Fraude", ascending=False)
)

display(device_info_fraud)
```

```python
# Gráfico

plt.figure(figsize=(10,6))

plt.barh(
    device_info_fraud.index.astype(str),
    device_info_fraud["Taxa_Fraude"]
)

plt.title("Taxa de Fraude dos Principais Dispositivos")
plt.xlabel("Taxa de Fraude (%)")

plt.tight_layout()
plt.show()
```

### Insight

A variável **DeviceType** apresentou uma das maiores capacidades discriminatórias observadas na EDA.

Transações realizadas em **dispositivos móveis** registraram taxa de fraude de **10,17%**, significativamente superior às transações realizadas em **desktops (6,52%)** e às transações sem informação de dispositivo (**2,10%**).

A análise de **DeviceInfo** mostrou que alguns dispositivos específicos apresentam taxas de fraude elevadas, porém muitas dessas categorias possuem baixo volume de observações. Entre os dispositivos com maior representatividade, **Windows** e **iOS Device** apresentaram taxas de fraude moderadamente elevadas, enquanto **MacOS** e **Trident/7.0** apresentaram risco inferior.

Esses resultados indicam que o contexto do dispositivo está fortemente associado ao risco de fraude e reforçam a importância de incluir **DeviceType** e versões consolidadas de **DeviceInfo** nas etapas de Engenharia de Atributos e modelagem preditiva.
